#### SET-UP PROJECT ENVIRONMENT FOR `ORACLE HRMS` DATA LAKEHOUSE 
##### CREATE STORAGE LOCATION FOR `ORACLE HRMS` CONTAINER
- CONTAINER NAME: hrms
- ACCOUNT NAME: oraclehrmsdevadls

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS oraclehrms_external_location
URL 'abfss://hrms@oraclehrmsdevadls.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `oraclehrms-dev-storage-credential`)
COMMENT 'CREATING EXTERNAL LOCATION FOR ORACLE HRMS CONTAINER'
```

```sql
%fs ls 'abfss://hrms@oraclehrmsdevadls.dfs.core.windows.net/'
```

```sql
SHOW CATALOGS;
```

#### CREATE CATALOG FOR `ORACLE HRMS` CONTAINER
- CATALOG NAME: ORACLE
- CONTAINER NAME: hrms

```sql
CREATE CATALOG IF NOT EXISTS oracle_hrms
MANAGED LOCATION 'abfss://hrms@oraclehrmsdevadls.dfs.core.windows.net/'
COMMENT 'CREATING CATALOG FOR ORACLE HRMS'
```

```sql
USE CATALOG oracle_hrms;
SELECT current_catalog();
```

#### CREATE VOLUME

```sql
USE CATALOG oracle_hrms;
USE SCHEMA landing;
CREATE EXTERNAL VOLUME IF NOT EXISTS operational
LOCATION 'abfss://hrms@oraclehrmsdevadls.dfs.core.windows.net/landing/operational'
COMMENT 'CREATE VOLUME opearation_data';
```
