### CREATE TABLE
- `ORACLE_HRMS.BRONZE.LOCATIONS`
- `ORACLE_HRMS.AUDIT.AUDIT_LOGS`

```sql
-- CREATE ORACLE_HRMS.BRONZE.LOCATIONS
CREATE TABLE IF NOT EXISTS ORACLE_HRMS.BRONZE.LOCATIONS
(
    location_id INT,
    location_name STRING,
    load_ts STRING,
    file_path STRING,
    file_name STRING,
    load_timestamp TIMESTAMP
)
```

```sql
-- CREATE AUDIT TABLE TO LOG OBSERVABILIY INFORMATION JOB RUN
CREATE TABLE IF NOT EXISTS ORACLE_HRMS.AUDIT.AUDIT_LOGS (
  log_id STRING,
  run_id STRING,
  event_time DATE,
  event_type STRING,
  source_table STRING,
  target_table STRING,
  record_count BIGINT,
  status STRING,
  message STRING,
  user_name STRING,
  notebook_path STRING,
  pipeline_name STRING,
  load_start_time TIMESTAMP,
  load_end_time TIMESTAMP
);
```
